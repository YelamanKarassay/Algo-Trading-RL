```python
from pathlib import Path
import pickle

import pandas as pd

path = Path("/Users/yela/Documents/University Programming/RL_quantphemes_2/results/exp_002_2800_30min_20260514_012414_118062/q_state.pkl")

with path.open("rb") as f:
    version = f.read(1)
    state = pickle.loads(f.read())

tables = state["tables"]
visits = state["visits"]

rows = []
for decision_point in range(tables.shape[0]):
    for state_id in range(tables.shape[1]):
        for action in range(tables.shape[2]):
            rows.append(
                {
                    "decision_point": decision_point,
                    "state": state_id,
                    "action": action,
                    "q_value": tables[decision_point, state_id, action],
                    "visits": visits[decision_point, state_id, action],
                }
            )

df = pd.DataFrame(rows)

print("version:", version)
display(df)
```
